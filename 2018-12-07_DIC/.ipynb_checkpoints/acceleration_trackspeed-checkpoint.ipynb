```python
import numpy as np
import matplotlib.pyplot as plt
import csv
import pandas as pd
import statistics as stat
```

```python
date = '2018-11-13'
ch = ['21','22','23','31','33']
conc_name = ['7-5','15','5','10','12-5','12-5','5','10']
conc = [7.5,15,5,10,12.5,12.5,5,10]
intr = [10,10,10,10,10,10,10,10]
```

```python
def acceleration(k):
    
    filename = date+'_CH'+ch[k]+'_'+conc_name[k]+'uM'
    
    data = pd.read_csv(filename+'.csv',sep=',');
    supp = pd.read_csv(filename+'-sup.csv',sep='\t');
    
    n=len(np.trim_zeros(supp['Tub flow']));
    inter = intr[k];
    
    start = sum(supp['Tub flow'])/n;
        
    s = abs(data['Track speed (um/s)'])*60;
    y1 = []
    
    for i in range(0,len(data['y1'])):
        y1 = np.append(y1,min(data['y1'][i],data['y2'][i]))
    
    track_start = (y1*inter - start)/60;
    speed = s.values
        
    return track_start,speed
```

```python
for i in range(0,len(ch)):
    
    track = acceleration(i)
    plt.scatter(track[0],track[1])

plt.legend(('7.5','15','5','10','12.5','12.5','5','10'),loc='upper right')
plt.xlabel('Track Start (min)')
plt.ylabel('Growth Rate ' r'$(\mu m / min)$')
plt.ylim(0,4)
plt.xlim(-0.5,20)
plt.grid(True)
plt.title('TIRF labeled 1:10');
```

```python
def growth_rate(k):
    
    filename = date+'_CH'+ch[k]+'_'+conc_name[k]+'uM'
    
    data = pd.read_csv(filename+'.csv',sep=',');
    supp = pd.read_csv(filename+'-sup.csv',sep='\t');
    
    utid = supp['uTub ID']
    uTubID = utid.tolist()
    
    IDs = set(uTubID) 
    speed = []
    
    for i in IDs:
        j = uTubID.index(i)
        s = abs(data['Track speed (um/s)'][j])*60;
        speed = np.append(speed,s)
    
    Mspeed = stat.mean(speed)
    Sspeed = stat.stdev(speed)
    
    return Mspeed, Sspeed
```

```python
m_speed = []
s_speed = []

for i in range(0,len(ch)):
    
    track = growth_rate(i)
    m_speed = np.append(m_speed,track[0])
    s_speed = np.append(s_speed,track[1])
```

```python
a1 , a0 = np.polyfit(conc, m_speed, 1)
y_int = -a0/a1
x_range = np.arange(0,15,0.1)

plt.errorbar(conc,m_speed, yerr=s_speed, fmt='o')
plt.plot(x_range, a1*x_range+a0)
plt.legend(('Linear Fit:  'r'$K_a$'' = %.2f 'r'$\frac{dimers}{\mu M \; s}$'', 'r'$C_c$'' = %.2f 'r'$\mu M$' %(a1*27.1, y_int),'Data'),loc='upper left')
plt.xlabel('[Tub] ' r'$(\mu M)$')
plt.ylabel('Growth Rate ' r'$(\mu m / min)$')
plt.ylim(0,2.5)
plt.xlim(0,16)
plt.grid(True)
plt.title('TIRF 1:10 labeled');
```

```python
nnuc = [8,0,12,0,0,5,9,2]
total = [36,12,25,6,29,18,35,36]
nuc = np.subtract(total,nnuc)
pnuc = np.divide(nuc,total)
conc = [7.5,15,5,10,12.5,12.5,5,10]
plt.scatter(conc,pnuc)
plt.xlabel('[Tub] ' r'$(\mu M)$')
plt.ylabel('P(nucleation) in t=15min')
plt.ylim(-0.1,1.1)
plt.xlim(0,16)
plt.grid(True)
plt.title('TIRF 1:10 labeled');
```
