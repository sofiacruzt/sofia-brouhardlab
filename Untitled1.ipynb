```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import csv
import pandas as pd
import statistics as stat
import scipy.stats as scistats
import matplotlib.mlab as mlab
from pathlib import Path
```

```python
date = '2019-05-02' # // 2019-01-31 // 2018-12-07
micscp = 'DIC' #TIRF or DIC

p = 'E:\\Sofia\\'+date+'_'+micscp
path = Path(p)
path.exists()
```

```python
ch = ['24']
conc_name = ['10']
conc = [10]

t_interval = 5
```

```python
def dynamics(k):
    
    filename = path/('CH'+ch[k]+'_'+conc_name[k]+'uM-results')/(date+'_CH'+ch[k]+'_'+conc_name[k]+'uM-1.csv')
    data = pd.read_csv(filename,sep=',');
    
    inter = t_interval;
    
    ####################################
    
    ID = [n-1 for n in data['Track ID']]
    growth = []
    lifetime = []
    
    for i in ID:
        g = abs(data['Track speed (um/s)'][i])*60;
        growth = np.append(growth,g)
       
        y2 = max(data['y1'][i],data['y2'][i])
        
        if y2 < 219:
            lt = (data['Track duration (s)'][i]);
            lifetime = np.append(lifetime,lt/60)
    
    ######################################
        
    return growth, lifetime
```

```python
font = {'family' : 'normal',
        'weight' : 'normal',
        'size'   : 12}
plt.rc('font', **font)

j=0
track = dynamics(j)
titles = ('Growth','Catastrophe')
xlables = ('Growth Rate ' r'$(\mu m / min)$','Lifetime ' r'$(min)$')

plt.figure(figsize=(15,5))
for i in [0,1] :
    
    plt.subplot(1,2,i+1)
    plt.hist(track[i],25, density=True)
    plt.ylabel('Relative Frequency')
    plt.xlabel(xlables[i])
    plt.title(titles[i]);

plt.suptitle('Histograms of [Tub]= %.2f' r'$\mu M$'%(conc[j]), fontsize=16)
```

```python
(mu0, sigma0) = scistats.norm.fit(track[0])
n, bins, patches = plt.hist(track[0], 20, density=True, alpha=0.75);
x = np.linspace(0, bins[-1], 100)
plt.plot(x, scistats.norm.pdf(x,mu0,sigma0),'--', lw=3, alpha=0.85)
legnd = ['10 uM, '+r'$\mu =$'+ str(round(mu0,3))+r', $\sigma =$'+ str(round(sigma0,3))]
plt.legend(legnd,loc='upper right')
```
