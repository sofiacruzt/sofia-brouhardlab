```python
import numpy as np
import matplotlib.pyplot as plt
import csv
import pandas as pd
from pathlib import Path
from read_roi import read_roi_zip
from collections import OrderedDict, Counter
import math
import PIL
from PIL import Image
from PIL.TiffTags import TAGS
import tkinter as tk
from tkinter import filedialog
import seaborn as sns
from pandas.api.types import CategoricalDtype
import itertools
from colour import Color
from datetime import date
import tifffile
from scipy import stats
from skimage import filters
```

```python
askdirectory = filedialog.askdirectory() # show an "Open" dialog box and return the path to the selected file
path = Path(askdirectory)
path
```

```python
composite=path/'2-_Composite_driftcorrected.tif'
composite
```

```python
im = tifffile.imread(composite)
im_dcx = im[0]
im_mts = im[1]
```

```python
val = filters.threshold_otsu(im_mts)
im_mts_mask = im_mts > val
im_dcx_masked = im_dcx * im_mts_mask
```

```python
fig, ax = plt.subplots(2,2,figsize=(20, 20))

ax[0,0].imshow(im_mts, interpolation='none', origin=[0, 0], cmap='viridis')
ax[0,1].imshow(im_mts_mask, interpolation='none', origin=[0, 0], cmap='gray')
ax[1,0].imshow(im_dcx, interpolation='none', origin=[0, 0], cmap='viridis')
ax[1,1].imshow(im_dcx_masked, interpolation='none', origin=[0, 0], cmap='gray')
```

```python
coeff, pval = stats.pearsonr(im_mts_mask.flatten(), im_dcx_masked.flatten())
coeff, pval
```
