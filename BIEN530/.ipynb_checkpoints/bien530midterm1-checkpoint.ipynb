```python
import numpy as np
import matplotlib.pyplot as plt
import csv
import pandas as pd
import statistics as stat
```

```python
path = 'C://Users/brouhardlab/Documents/GitHub/sofia-brouhardlab/BIEN530/'
filename = path+'spots_stats.csv'
data = pd.read_csv(filename,sep=',');
```

```python
def firstnlast(data):
    ID =data.iloc[[-1]]['TRACK_ID']
    ID = ID.values
    
    fnl_data = data.loc[(data['TRACK_ID'] == 0)]
    fnl_data = fnl_data.iloc[[0, -1]]
    
    for i in range(1,ID[0]):
        df = data.loc[(data['TRACK_ID'] == i)]
        df = df.iloc[[0, -1]]
        fnl_data = pd.concat([fnl_data, df], sort=False)
        
    df.to_csv(path+'spots_stats_fnl.csv', encoding='utf-8', index=False)
```
