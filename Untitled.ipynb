```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import csv
import pandas as pd
import statistics as stat
from pathlib import Path
```

```python
date = '2018-12-07' # // 2019-01-31 // 2018-12-07
micscp = 'DIC' #TIRF or DIC

p = 'C:\\Users\\brouhardlab\Documents\GitHub\sofia-brouhardlab\\'+date+'_'+micscp
path = Path(p)
path.exists()
```

```python
data_nuc = pd.read_csv(path/(date+'_time2nuc.csv'),sep=',');
data_cat = pd.read_csv(path/(date+'_lifetime.csv'),sep=',');
```

```python
df = pd.DataFrame( columns=['Concentration (uM)','Lag (min)'])

for i in list(data_nuc):
    l = len(data_nuc[i].dropna())
    c = float(i)*np.ones(l)
    df0 = pd.DataFrame( columns=['Concentration (uM)','Lag (min)'])
    df0['Concentration (uM)'] = c
    df0['Lag (min)'] = data_nuc[i].dropna()
    
    df = df.append(df0)
```

```python
df1 = pd.DataFrame( columns=['Concentration (uM)','Lag (min)'])

for i in list(data_cat):
    l = len(data_cat[i].dropna())
    c = float(i)*np.ones(l)
    df0 = pd.DataFrame( columns=['Concentration (uM)','Lag (min)'])
    df0['Concentration (uM)'] = c
    df0['Lag (min)'] = data_cat[i].dropna()
    
    df1 = df1.append(df0)
```

```python
data = pd.read_csv(path/(date+'_data.csv'),sep=',', index_col=0);
data
```

```python
titles = ('Nucleation','Catastrophe', 'Correlation?')
ylables = ('Lag ' r'$(min)$','Lifetime ' r'$(min)$', 'Lag ' r'$(min)$')
xlables = ('[Tub] ' r'$(\mu M)$','[Tub] ' r'$(\mu M)$','Lifetime ' r'$(min)$')

font = {'family' : 'normal',
        'weight' : 'normal',
        'size'   : 18}
plt.rc('font', **font)

plt.figure(figsize=(20,5))
plt.suptitle('Analysis of '+micscp+' data from '+date, fontsize=16)

plt.subplot(1,3,1)
plt.errorbar(data['tub conc'], data['time2nuc mean'], yerr=data['time2nuc std'], fmt='o', markersize=0, c='black')
plt.scatter(data['tub conc'], data['time2nuc mean'],c=['#1F77B4','#FF7F0E','#E60A85','#0CBCE5','#26C400'],s=200)
plt.ylim(0,11)

plt.subplot(1,3,2)
plt.errorbar(data['tub conc'], data['lifetime mean'], yerr=data['lifetime std'], fmt='o', markersize=0, c='black')
plt.scatter(data['tub conc'], data['lifetime mean'],c=['#1F77B4','#FF7F0E','#E60A85','#0CBCE5','#26C400'],s=200)
plt.ylim(0,21)

plt.subplot(1,3,3)
plt.errorbar(data['lifetime mean'], data['time2nuc mean'], yerr=data['time2nuc std'], xerr=data['lifetime std'], fmt='o', markersize=0, c='black')
plt.scatter(data['lifetime mean'], data['time2nuc mean'],c=['#1F77B4','#FF7F0E','#E60A85','#0CBCE5','#26C400'],s=200)
plt.ylim(0,11)


for i in [0,1,2] :
    plt.subplot(1,3,i+1)
    plt.xlabel(xlables[i])
    plt.ylabel(ylables[i])
    plt.title(titles[i]);
    plt.legend(( 'Mean','S Dev'),loc='upper left')
    plt.xlim(0,21)
    plt.grid(True)
    
plt.savefig(path/(date+'_graphs.pdf'))
plt.savefig(path/(date+'_graphs.png'))
```
