```python
#Geting opening relevant file
askdirectory=filedialog.askdirectory()
path=Path(askdirectory)
path
indiv_data_df=pd.read_csv(path/'Sofia_2020-08-03.csv',sep=",")
indiv_data_df
```

```python
tubulin = '[Tubulin] ' r'$(\mu M)$'
GrowthRate = 'Growth Rate ' r'$(\mu m / min)$'
g=sns.FacetGrid(indiv_data_df,row=tubulin, col='DCX Type', margin_titles=True)
bins=np.linspace(0,1.5,20)
g.map(plt.hist, GrowthRate, bins=bins, density=True)
#making the first histogram showing the relation between DCX type and growth rate and tubulin and growth rate 
```
