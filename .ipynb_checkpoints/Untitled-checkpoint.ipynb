```python
import matplotlib.pyplot as plt
from numpy import exp, loadtxt, pi, sqrt

from lmfit import Model

data = loadtxt('model1d_gauss.dat')
x = data[:, 0]
y = data[:, 1]


def gaussian(x, amp, cen, wid):
    """1-d gaussian: gaussian(x, amp, cen, wid)"""
    return (amp / (sqrt(2*pi) * wid)) * exp(-(x-cen)**2 / (2*wid**2))


gmodel = Model(gaussian)
result = gmodel.fit(y, x=x, amp=5, cen=5, wid=1)

print(result.fit_report())

plt.plot(x, y, 'bo')
plt.plot(x, result.init_fit, 'k--')
plt.plot(x, result.best_fit, 'r-')
plt.show()
```

```python
import numpy as np
from scipy.optimize import curve_fit
import matplotlib.pyplot as plot


def gaussian(x, mu, sig):
    return (np.exp(-np.power(x - mu, 2.) / (2 * np.power(sig, 2.))) )/(sig*np.sqrt(2*np.pi))

y = np.random.normal(1,0.5,100)
n, bins, patches = plt.hist(y, bins=np.arange(-3, 3 + 0.05, 0.05), density=True, alpha=0.65);

coeff, var_matrix = curve_fit(gaussian,bins[:-1],n)
variance = np.diagonal(var_matrix) #Refer [3]

SE = np.sqrt(variance) #Refer [4]

#======Making a dictionary to print results========
r = {'mu':[coeff[0],SE[0]],'sigma':[coeff[1],SE[1]]}
results = pd.DataFrame(data=r, index=(['Value','SE']))

#========End Results Printing=================

y2 = gaussian(bins[:-1],coeff[0],coeff[1]) #Saves the y values for the fitted model

plot.plot(bins[:-1],y2)

plot.show()

print(results)
```

```python
def gaussian_fit(data):
    r = {'[tub]':np.empty(0),'mu':np.empty(0) ,'sigma':np.empty(0)}
    results = pd.DataFrame(data=r)
    
    for i in list(data):
        y = data[i].dropna().values
        
        n, bins, patches = plt.hist(y, bins=np.arange(0, 3 + 0.05, 0.05), density=True);
        plt.clf()

        coeff, var_matrix = curve_fit(gaussian,bins[:-1],n)
        
        variance = np.diagonal(var_matrix) #Refer [3]
        SE = np.sqrt(variance) #Refer [4]

#======Making a data frame========
        r = {'[tub]':[float(i),float(i)],'mu':[coeff[0],SE[0]],'sigma':[coeff[1],SE[1]]}
        results0 = pd.DataFrame(data=r, index=(['Value','SE']))
        results = pd.concat([results, results0], sort=False)
    
    return results
```

```python
def exponential(x, scale):
    return ((np.exp(-x/scale) )/(scale))
```

```python

datos = dataN['8'].dropna().values

y = datos
n, bins, patches = plt.hist(y, bins=np.arange(0, 60 + 0.5, 0.5), density=True, alpha=0.65);

coeff, var_matrix = curve_fit(exponential,bins[:-1],n)
variance = np.diagonal(var_matrix) #Refer [3]

SE = np.sqrt(variance) #Refer [4]

#======Making a dictionary to print results========
r = {'scale':[coeff[0],SE[0]]}
results = pd.DataFrame(data=r, index=(['Value','SE']))

#========End Results Printing=================

y2 = exponential(bins[:-1],coeff[0]) #Saves the y values for the fitted model

plot.plot(bins[:-1],y2)

plot.show()

print(results)
```

```python
def gamma(x, shape, scale):
    return (np.power(x,shape-1)*np.exp(-x/ scale))/(scipy.special.gamma(shape) * np.power(scale,shape))
```

```python
datos = dataC['15'].dropna().values

y = datos
n, bins, patches = plt.hist(y, bins=np.arange(0, 60 + 0.5, 0.5), density=True, alpha=0.65);

coeff, var_matrix = curve_fit(gamma,bins[:-1],n,[1.1,1])
variance = np.diagonal(var_matrix) #Refer [3]

SE = np.sqrt(variance) #Refer [4]

#======Making a dictionary to print results========
r = {'shape':[coeff[0],SE[0]],'scale':[coeff[1],SE[1]]}
results = pd.DataFrame(data=r, index=(['Value','SE']))

#========End Results Printing=================

y2 = gamma(bins[:-1],coeff[0],coeff[1]) #Saves the y values for the fitted model

plot.plot(bins[:-1],y2)

plot.show()

print(results)
```
